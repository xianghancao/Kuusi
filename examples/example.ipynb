# Kuusi Demo
![Kuusi overview](../docs/assets/Logo-1.png)

One notebook, two ways to see it: **Article** for reading and printing, **Mind Map** for structure.

## Start here

Click **Kuusi View** in the toolbar (or **Open With** a Kuusi factory), then switch **Article / Mind Map** in the view header.

Headings become mind-map nodes; cells under a heading attach to that topic. `#` is the root, deeper headings nest, and non-heading cells belong to the nearest heading above.

### Mind Map keys

- **Click** select · **Double-click / F2** edit
- **Enter** sibling · **Tab** child · drag the left handle to reorder
- Wheel pan · **Ctrl/Cmd + wheel** zoom · **Shift+1** fit the whole map
- Header menus: **Theme · Text · Paragraph · Elements · Layout · Node · Line**

### Article menus

**Theme · Page · Text · Paragraph · Elements**, plus **Export** for PDF, PNG, print and HTML.

Try these in **Page**:

- **Contents** — a table of contents with page numbers, and an optional cover page
- **Chapters** — number headings (`1`, `1.1`, `1.1.1`) and start each chapter on a new page
- **Paper** and **Margins** — A4, A5, Letter … with per-side margins and a binding gutter
- **Header / Footer** — `{title}` `{chapter}` `{date}` `{page}`, rule lines, different first page
- **Watermark** — diagonal text such as DRAFT

Exported PDFs carry one bookmark per heading.

## Worked example: a short report

The sections below read like an ordinary report. They are here on purpose so the document runs over several pages: try **Page → Contents** for a contents page, **Header / Footer** for running heads and page numbers, and **Page → Chapters** to start every second-level heading on a new page.

The figures are made up for illustration; the point is how long-form text, lists, tables and code flow across pages.

### Background

A notebook is a good place to think, but a poor place to hand work to someone else. Cells are ordered by when they were written, not by what a reader needs first, and the surrounding interface is built for editing rather than reading. Teams usually solve this at the end, by copying the useful parts into a separate document.

That copy step is where the trouble starts. The text drifts away from the notebook, figures are pasted as screenshots, and nobody remembers which version of the analysis a paragraph came from. Keeping one source and rendering it in the shape each audience needs removes the copy altogether.

### Method

We wrote the report once, as Markdown cells under a clear heading structure, and looked at the same file in two ways. The outline view was used while planning and reordering; the page view was used while reading, proofreading and printing. For each draft we followed the same routine:

1. Sketch the headings in the outline and move them until the argument read in order.
2. Fill each heading with a few short paragraphs, adding a table or a code cell only where it carried evidence.
3. Switch to the page view, check the page breaks, and fix anything that split awkwardly.
4. Export a PDF and read it once on paper before sharing.

Nothing was copied between tools at any step, so every export reflected the latest text.

### Results

Across ten drafts the routine was quick to follow. Most of the saved time came from not rebuilding the document by hand at the end, and the page view caught layout problems while they were still cheap to fix. The table summarises what changed (illustrative numbers).

| Measure | Before | After | Change |
| --- | --- | --- | --- |
| Time to a shareable draft | 95 min | 60 min | −37% |
| Manual copy steps per draft | 6 | 0 | −100% |
| Layout fixes found after export | 4 | 1 | −75% |
| Drafts out of sync with the notebook | 3 of 10 | 0 of 10 | −100% |

The smallest improvement was in proofreading time, which stayed roughly flat: reading carefully takes as long as it always did.

```python
before = {"draft": 95, "copy_steps": 6, "layout_fixes": 4}
after = {"draft": 60, "copy_steps": 0, "layout_fixes": 1}

for key in before:
    change = (after[key] - before[key]) / before[key]
    print(f"{key:<13} {before[key]:>3} -> {after[key]:>3}  ({change:+.0%})")
```

### Discussion

Two things made the difference. First, the heading structure did double duty: it was the outline while planning and the table of contents while reading, so it never had to be maintained twice. Second, keeping the writing in plain Markdown meant the text could be reviewed, diffed and edited anywhere, while the page layout stayed a separate, adjustable choice.

There are limits worth stating plainly:

- Very wide tables still need a narrower font or a landscape page.
- Long code cells split across pages, so keep examples short or turn on line numbers.
- Footnotes are generated from links, not written by hand, so citations still belong in the text.

> A document should be written once and shown in the form each reader needs.

None of this removes the need for editing. It only removes the busywork around it.

### Conclusion

Treating the notebook as the single source, and the outline and page views as two ways of looking at it, shortened the path from idea to a shareable document. The approach is cheap to try: write under headings, check the page view early, and export only when the text is ready.
