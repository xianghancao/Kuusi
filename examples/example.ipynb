# Kuusi Demo
![Kuusi overview](../docs/assets/Logo-1.png)

Click **Kuusi View** in the toolbar (or **Open With** a Kuusi factory).

Switch **Article / Mind Map / Slides**. Headings become mind-map nodes; cells under a heading attach to that topic.


## Try these


- **Click** select · **Double-click / F2** edit
- **Enter** sibling · **Tab** child · drag left handle to reorder
- Wheel pan · **Ctrl/Cmd + wheel** zoom
- Mind Map header: **Theme · Text · Paragraph · Headings · Code · Image · Line · Node · Background**
- Article / Slides: Theme · Page · Header & footer · Text · Image; **Export** for PDF or PNG


## Outline


`#` is the root. Deeper headings nest. Non-heading cells belong to the nearest heading above.
