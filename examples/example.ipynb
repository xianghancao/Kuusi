# Kuusi Demo


Open with **Kuusi** in the toolbar (or **Open With → Kuusi Mind Map**).

Headings become nodes; cells under a heading attach to that topic.


## Try these


- **Click** select · **Double-click / F2** edit
- **Enter** sibling · **Tab** child · drag left handle to reorder
- Wheel pan · **Ctrl/Cmd + wheel** zoom
- Header: **Theme / Font / Line / Node / Background / Layout**


## Outline


`#` is the root. Deeper headings nest. Non-heading cells belong to the nearest heading above.
